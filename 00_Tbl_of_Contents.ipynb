# Table of Contents

This set of notebooks contains code and data for exploring the cycles of gaps $\mathcal{G}(p^\#)$ that occur at each stage of Eratosthenes sieve.

### 0x - Cycles of gaps - the dynamics
<i>01_Cycles</i>  : Code for the 3-step recursion to create cycles of gaps $\mathcal{G}(p_k^\#)\longrightarrow \mathcal{G}(p_{k+1}^\#)$. This code works well up through $p=29$.  The files G11uint, G17uint, and G19uint are included for convenience.

<i>02_p-rough_interactive</i> : Code for exploring the $29$-rough numbers through an interactive graph of $\Delta \Phi(x,29)$.

<i>03_primeblock</i> : Code for finding the prime numbers in an interval.  We provide a starting point $x_0$ for the search
and a blocklength.  We only search over the odd numbers, so the interval is $[x_0, x_0+2\cdot blocklength]$.
There is also a function for extending an existing block of primes.

### 1x - Population models for gaps
<i>10_wg</i> :  Code for calculating and displaying the relative population models $w_{g,1}(\lambda)$ for $g \le 82$.

Data file <i>lj37</i>:  contains the coefficients for $w_{g,j}(\lambda)$ for gaps $2 \le g \le 82$ and using $p_0=37$
for initial conditions.

<i>11_lambda</i> : Code for calculating the value of the parameter $\lambda = \prod_{41}^{p_k} \frac{p-3}{p-2}$ with $p_0=37$, across the span of $\mathcal{G}(23^\#)$.

<i>12_DelHprep</i> : Code for calculating the accumulated gaps across consecutive intervals of survival $\Delta H(p)$.  Includes code
for combining adjacent blocks of counts over intervals of survival.

Data files <i>DHExx_nnn</i>:  Data files of counts of gaps $g$ over $\Delta H(p)$ with $p^2 \approx n.nn Exx$

<i>13_DelHdisplay</i> : Code for displaying the accumulated gaps across consecutive intervals of survival $\Delta H(p)$ and comparing
these counts to estimates based on the models $w_{g,1}(\lambda)$.

<i>16_bias_lastdigits</i> : Code for understanding and exploring the biases observed by Lemke Oliver and Soundararajan in
their 2016 statistical study of the last digits of consecutive primes.  With the relative population models 
$w_{g,1}(\lambda)$ we see that the observed biases are entirely expected and transient, although these biases will
persist well beyond the computational range.

### 2x - Population models for constellations

<i>a20s_class.py</i> : include file for defining dictionaries of constellations of interest, by length.  Arrays of keys, setting an order for 
processing and displaying the constellations.

<i>20_ws</i> : Code for calculating and displaying the relative population models $w_{s,J}(\lambda)$ for the constellations
listed in a20s_class.py.

Data files <i>J3lj.pkl</i> and <i>J5lj.pkl</i>:  coefficients for the relative population models $w_{s,J}(\lambda)$ for 
the constellations in <i>a20s_class.py</i> of length $J=3$ and $J=5$.

<i>21_DelHsprep</i> : Code for counting the occurrences of constellations across consecutive intervals of survival $\Delta H(p)$.

Data files <i>DHsExx_nnn</i>:  Data files of counts of constellations $s$ over $\Delta H(p)$ with $p^2 \approx n.nn Exx$

<i>22_DelHs_display</i> : Code for displaying the numbers of occurrences of the constellations across consecutive intervals of survival
$\Delta H(p)$ and comparing these counts to estimates based on the models $w_{s,J}(\lambda)$.

<i>23_nonconvex_prep_Engelsma459</i> : Code for tracking the 58 Engelsma $(459,3242)$-counterexamples $s$ 
to the convexity conjecture from their (inadmissible) 
driving term in ${\mathcal G}(11^\#)$, through the occurrence of $s$ in ${\mathcal G}(113^\#)$, and up to the
end of each unique prefix of primorial coordinates.  
This notebook includes an <b> interactive graph</b> of each of the 58 counterexamples compared to the constellation of
small primes.  The outputs include files of the prefixes and constellations sorted by their prefixes, and files
summarizing extensions of these prefixes in a breadth-first search.  We record those extensions that end in one or
more coordinates $m_k=0$.  Read the comments in the file for description of these files.

<i>24_nonconvex_search_Engelsma459</i> : Code for probing for a survivor among the Engelsma $(459,3242)$-counterexamples.  While the
initial search in "23_nonconvex..." is breadth-first and exhaustive, the code here switches to opportunistic depth-first searches, 
following strings of $m_k=0$ in a row.

<i>25_nonconvex_prep_Engelsma458</i> : Code for tracking the 116 Engelsma $(458,3240)$-counterexamples to the convexity
conjecture.  This notebook is an adaptation of <i>23_nonconvex ...</i> for the $(458,3240)$ constellations.


### 3x - studies on surviving the sieve
<i>33_quaddensity</i> : 
This code produces an interactive plot of the quadratic density 
$$\eta(p_k) = \frac{1}{p_{k+1}-p_k} N_{\Delta H}(p_k)$$
over the intervals of survival and shows the mean and standard deviation for each column.  
The quadratic density provides an average occurrence of a gap $g$
over intervals $[n^2,(n+1)^2]$.

<i>34_wgj_evolution</i>:
This notebook produces an interactive heatmap that illustrates the evolution of the relative populations
$w_{g,j}(p^\#)$ of gaps $2 \le g \le 82$ and their
driving terms across stages of the sieve.  The heatmap is shown across $j$ and $g$, and the asymptotic relative populations
are shown in the first column.  There is a button for switching the heatmap to see 
$\delta_{|s|,j}(\lambda) = w_{g,j+1}(\lambda)/w_{g,j}(\lambda)$

### 9x - notes
<i> 90_notes</i> : Notes on this project.  Tables of constants, lengths and sizes of cycles ${\mathcal G}(p^\#)$,
horizons of survival, etc.
