# Notes on the Cycles of gaps
At every stage of Eratosthenes sieve there is a cycle of gaps $\mathcal{G}(p^\#)$ between the $p$-rough numbers.  

The cycle $\mathcal{G}(p^\#)$ has length $\phi(p^\#)$ gaps of span $p^\#$.  

The cycle $\mathcal{G}(p^\#)$ is symmetric: $g_{\phi(p^\#)-i} = g_i$, and $g_{\phi(p^\#)}=2$.

## Notes on the p-rough numbers
The enumeration $\Phi(x,p)$ of p-rough numbers is completely determined by the differences
$$ \Delta \Phi(x,p) = \Phi(x,p) - \frac{1}{\mu}x$$
over one period of the cycle $\mathcal{G}(p^\#)$.  $\Delta \Phi(x,p)$ is bounded and periodic, and it has rotational
symmetries around the waypoints $x_{k,1}=k\cdot p^\#$ and $x_{k,2} = (k-\frac{1}{2}) \cdot p^\#$.
$$ \Delta \Phi(x_{k,1},p) = \Delta \Phi(x_{k,2},p) = 0 $$

For a few early primes we tabulate the maximum difference of $\Phi(x,p)$ away from its line of symmetry,
the number $N_0^+$ of zeroes for $\Delta \Phi(x,p)$ on its vertical segments in one cycle, and the proportion of these
among all the vertical segments.

## Constant for Mertens' approximation
To associate primes with values of the system parameter $\lambda$ in $w_{g,1}(\lambda) = w_{g,1}(p^\#)$, we rely on <b>Mertens' Third Theorem</b>.
$$ \prod_2^p \frac{q-1}{q} = \frac{e^{-\gamma}+ o(1)}{\ln p} $$
In studying $w_{g,1}(p^\#)$ we start from a cycle ${\mathcal G}(p_0^\#)$ and have the parameter
$$ \lambda(p_k) \; = \; \prod_{p_1}^{p_k} \frac{q-2}{q-1} \; \approx \frac{c}{\ln p_k}$$

From the data for $\lambda$ calculated for $p_0=37$ we estimate that $c = 3.7323679667507315$

## Files Engjjj_xx_results.csv and Engjjj_xx_mzero.csv
The files <i>Engjjj_xx_results.csv</i> store the results from the notebook <i>23_nonconvex..</i> breadth-first search for admissible instances 
of the Engelsma counterexamples of length $J=jjj$ (e.g. $459$).  The index $xx$ identifies the specific counterexample.

Fields:

#1. $xx$: the identifier of the specific counterexample of length $J=jjj$

#2. <i>len_pref</i>: the length of the unique prefix of primorial coordinates, starting with $\gamma_0$ in ${\mathcal G}(11^\#)$.

#3. $p_k$: the first prime <i>beyond</i> the unique prefix.  $p_k$ is the first prime with $q-\nu_q(s)>1$.

#4. $k_{1}$: the depth of the breadth-first search, starting after $p_k$. The prefix is extended by $k_1+1$ coordinates $m_k$.

#5. $p_{k_{1}}$: the prime corresponding to the last primorial coordinate $m_k$ in the search.

#6. $N$: the number of admissible instances searched.

#7. array of $(q-\nu_q(s))$, the number of admissible instances for each prime $q$, starting at $p_k$.  This array runs far past $p_{k_1}$.



#8. The two factors in $w_s(\infty)$, the length factor and the span factor,
$$\prod_{q\le J+1}(q-\nu_q) \; \; \cdot \; \prod_{J+1 < q,\, q|Q} \frac{q-\nu_q}{q-J-1}$$
